- *¿Cuáles son las fases de la etapa de Análisis (Front-End) de un compilador?*  
**RESPUESTA:**
La etapa de Análisis (Front-End) de un compilador se encarga de entender el programa fuente, verificar que esté bien escrito y estructurado (tanto en sintaxis como en semántica), y traducir ese código a una representación intermedia (IR) independiente de la máquina de destino
. Al concentrar los detalles del lenguaje de origen en esta etapa, se facilita la reutilización del Front-End para construir compiladores dirigidos a otras plataformas
.
Esta etapa se compone de cuatro fases principales
:
1. Análisis Léxico (o Lexicográfico / Escaneo)
Función: Es la primera fase del compilador
. Consume el flujo de caracteres del programa de izquierda a derecha y los agrupa en secuencias con significado propio llamadas componentes léxicos o tokens (como palabras reservadas, identificadores de variables, constantes numéricas y operadores)
.
Tareas adicionales: Durante este escaneo se eliminan los elementos que no aportan valor a la traducción, tales como comentarios, espacios en blanco innecesarios, tabuladores y retornos de carro
.
2. Análisis Sintáctico (o Parser)
Función: Toma la secuencia de tokens provista por el analizador léxico y comprueba si la estructura cumple con las reglas gramaticales (la gramática libre de contexto) del lenguaje de programación
.
Resultado: Genera una estructura jerárquica que representa la composición gramatical del programa, comúnmente en forma de un árbol sintáctico o un árbol sintáctico abstracto (AST)
.
3. Análisis Semántico
Función: Revisa el árbol sintáctico y la información de los símbolos para comprobar que el programa respete las reglas del lenguaje que van más allá de la estructura sintáctica pura, enfocándose en el significado de las operaciones
.
Tareas clave: Su componente más crítico es la verificación o comprobación de tipos (type checking), donde el compilador se asegura de que los operadores se apliquen a operandos compatibles (por ejemplo, impidiendo sumar un texto con un booleano, o gestionando conversiones automáticas de entero a real cuando el lenguaje lo permita)
. También verifica que las variables hayan sido declaradas antes de usarse, que estén dentro de sus ámbitos (scopes) correctos y que los parámetros en las llamadas a funciones coincidan en tipo y cantidad
.
4. Generación de Código Intermedio
Función: Una vez que se ha comprobado la total corrección del programa, se traduce el árbol sintáctico a una representación de bajo nivel pero abstracta, diseñada para una máquina virtual o inexistente (como el código de tres direcciones o cuádruplas)
.
Utilidad: Este código intermedio debe ser fácil de generar y de traducir al lenguaje de la máquina destino, permitiendo además la aplicación de optimizaciones independientes de la plataforma física
.
Componentes Transversales del Front-End
Para que estas fases funcionen de forma integrada, interactúan con dos estructuras fundamentales:
Tabla de Símbolos: Es una estructura de datos (frecuentemente una tabla hash o un árbol) que almacena información relevante sobre los identificadores declarados (variables, funciones, tipos de datos, etc.)
. Se inicializa y enriquece principalmente en las fases de análisis y es accedida constantemente por todo el compilador
.
Manejador de Errores: Cada una de las fases de análisis debe estar preparada para detectar errores del programador (léxicos, sintácticos o semánticos), reportar mensajes informativos y aplicar técnicas de recuperación de errores para continuar analizando el resto del archivo


- *¿Qué diferencia a un Lexema de un Token?*  
**RESPUESTA:**
En la construcción de compiladores, los conceptos de lexema y token se encuentran estrechamente relacionados en el proceso de análisis, pero representan diferentes niveles de abstracción
:
1. El Lexema (La cadena de caracteres real)
Un lexema es la secuencia concreta de caracteres en el texto del código fuente que coincide con un patrón determinado y es identificada por el analizador léxico como una unidad lingüística
.
Es textual y físico: Se trata del texto exacto tal y como lo escribió el programador en su archivo de código.
Ejemplos:
Las cadenas de texto "x", "posicion" o "puntuacion" escritas en el código
.
Las secuencias de dígitos de un número como "60" o "3.1416"
.
La secuencia exacta de caracteres de una palabra clave como "while" o "if"
.
2. El Token (La categoría abstracta)
Un token es un símbolo abstracto (o categoría gramatical) que representa a una clase o familia de componentes léxicos
. Es el elemento con el que realmente trabaja el analizador sintáctico (parser) para validar si la estructura gramatical del programa es correcta
.
Es lógico y numérico: Internamente, para el compilador es ineficiente procesar texto plano constantemente. Por ello, el analizador léxico traduce los lexemas y representa los tokens mediante números enteros únicos o tipos enumerados
.
Estructura típica: Se suele entregar al analizador sintáctico como un par de la forma 〈nombre-token, valor-atributo〉, donde el nombre-token es la categoría gramatical abstracta (como id o numero) y el valor-atributo es un puntero o dato adicional (por ejemplo, la entrada en la tabla de símbolos para recuperar el nombre real de la variable o el valor numérico)
.
Diferencias Clave y Relación
Para entender la diferencia de forma clara, se pueden observar los siguientes aspectos:
La analogía de la categoría y el objeto: El token es como la palabra genérica "Fruta", mientras que los lexemas son las frutas concretas como "manzana", "plátano" o "uva"
. Donde la gramática de un lenguaje requiera el token "Fruta", se podrá colocar cualquiera de sus lexemas asociados
.
No siempre es una relación uno a uno:
Tokens no específicos (Genéricos): Un token como id (identificador) o numero representa una categoría que engloba a infinitos lexemas posibles (por ejemplo, los lexemas "miVariable", "temperatura", "contador" se mapean todos al mismo token id)
. El atributo del token es lo que permite al compilador diferenciar un lexema de otro en etapas posteriores
.
Tokens específicos: Por el contrario, existen tokens para palabras reservadas (como while o if) u operadores específicos (como +) que habitualmente coinciden únicamente con un solo lexema posible ("while", "if" o "+", respectivamente)

## El "Monstruo Completo" (Enlace Estático)  
Por defecto, GCC usa ***enlace dinámico*** usando una estructura de datos tabular, la tabla **PLT** (*Procedure Linkage Table*) ya que el código usa `printf`, el programa necesita conectarse con la biblioteca estándar de C (`libc.so`). El enlazador no copia el código de `printf` dentro del archivo para no duplicar espacio en el disco; en su lugar, crea un "trampolín" o acceso directo en una sección llamada `<printf@plt>`.

Cada vez que se invoca a `printf`, realmente salta a este bloque plt, el cual averigua en qué parte de la memoria RAM del sistema operativo se encuentra la verdadera función `printf` y redirige el control allí en tiempo de ejecución.

Para ver cómo el enlazador inyecta literalmente miles y miles de líneas copiando físicamente el código de `printf` y todas sus dependencias dentro del ejecutable, se debe compilar de forma estática:

````python
import token
import tokenize
from io import BytesIO

# Código fuente de prueba en formato cadena
codigo_fuente = "suma = a + 10"

# Tokenización del código fuente
tokens = tokenize.tokenize(BytesIO(codigo_fuente.encode("utf-8")).readline)

print(
    f"{'LINEA/COL':<12} | {'TIPO DE TOKEN':<20} | {'VALOR (LEXEMA)':<15}"
)
print("-" * 55)
for tok in tokens:
    if tok.type in (
        tokenize.ENCODING,
        tokenize.ENDMARKER,
        tokenize.NL,
        tokenize.NEWLINE,
    ):
        continue
    nombre_token = token.tok_name[tok.type]
    posicion = f"{tok.start[0]}:{tok.start[1]}"
    print(f"{posicion:<12} | {nombre_token:<20} | {tok.string:<15}")
````

### [CÓDIGO 3.1] Simulación de una Tabla de Símbolos básica en Python
```python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos

ts.insertar("a", "ENTERO", 5)

```


````python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos
ts = TablaDeSimbolos()
ts.insertar("a", "ENTERO", 5)
ts.insertar("b", "ENTERO", 10)
ts.insertar("suma", "ENTERO", 15)
ts.mostrar()
````

````python
import re

# Redefine RESERVADAS, TOKENS_DEFINITIONS, REGEX for self-containment and to ensure correctness
# (The previous tokenizar definition in cell eo1dTaRevITn had a syntax error and did not return tokens).
RESERVADAS = {
    "DISPOSITIVO", "SENSOR", "PARAMETRO", "SI", "SINO", "MIENTRAS",
    "ENCENDER", "APAGAR", "ESTADO", "CUANDO", "Y", "O", "NO",
    "VERDADERO", "FALSO"
}

TOKENS_DEFINITIONS = [
    ("FLOTANTE", r"\d+\.\d+"), ("ENTERO", r"\d+"),
    ("CADENA", r'"[^"\n]*"'), ("FLECHA", r"->"),
    ("REL_OP", r">=|<=|==|!=|>|<"), ("ARIT_OP", r"[+\-*/]"),
    ("ASIGNACION", r"="), ("LLAVE_ABRE", r"\{"), ("LLAVE_CIERRA", r"\}"),
    ("IDENTIFICADOR", r"[A-Za-z_][A-Za-z0-9_]*"),
    ("ESPACIO", r"[ \t]+"), ("SALTO_LINEA", r"\n"), ("DESCONOCIDO", r".")
]

REGEX = "|".join(f"(?P<{n}>{p})" for n, p in TOKENS_DEFINITIONS)

def tokenizar(codigo):
    tokens_list, linea = [], 1

    for m in re.finditer(REGEX, codigo):
        tipo, valor = m.lastgroup, m.group()

        if tipo == "SALTO_LINEA":
            linea += 1
        elif tipo != "ESPACIO":
            if tipo == "IDENTIFICADOR" and valor in RESERVADAS:
                tipo = "PALABRA_RESERVADA"
            tokens_list.append({"tipo": tipo, "valor": valor, "linea": linea})
    return tokens_list

# Read the content of the example script from the file
with open("homescript_ejemplo.txt", "r") as f:
    codigo_fuente = f.read()

# Generate the tokens list by calling the tokenizar function
tokens = tokenizar(codigo_fuente)

tabla_simbolos = {}

def registrar_simbolo(tokens_param):
    for i, token_item in enumerate(tokens_param):
        if token_item["tipo"] == "PALABRA_RESERVADA" and token_item["valor"] in ("DISPOSITIVO", "SENSOR", "PARAMETRO"):
            categoria = token_item["valor"]
            # Assuming the next token is the name of the entity/parameter
            nombre = tokens_param[i + 1]["valor"]

            entrada = {
                "categoria": categoria,
                "tipo_dato": "entidad", # Default type
                "valor_inicial": None,
                "ambito": "global",
                "linea_declaracion": token_item["linea"]
            }

            # Special handling for PARAMETRO with assignment (e.g., PARAMETRO frio = 18)
            if (categoria == "PARAMETRO" and i + 3 < len(tokens_param)
                    and tokens_param[i + 2]["tipo"] == "ASIGNACION"):
                valor_token = tokens_param[i + 3]
                entrada["valor_inicial"] = valor_token["valor"]
                entrada["tipo_dato"] = "flotante" if valor_token["tipo"] == "FLOTANTE" else "entero"

            tabla_simbolos[nombre] = entrada

# Call registrar_simbolo with the generated tokens list
registrar_simbolo(tokens)

# Print the symbol table
for nombre, datos in tabla_simbolos.items():
    print(nombre, "->", datos)
````
